## Adım 1 · Tahmin Et

Bu kodu **çalıştırmadan önce** ekrana ne yazacağını çalışma kâğıdına yaz. Öğretmen söyleyince (Adım 2) eşinle birlikte çalıştır ve tahmininle karşılaştır.

```python
for i in range(1, 5):
    print(i, "x 3 =", i * 3)
```

## Adım 3 · İzle

1. **Önce tahmin et:** Aşağıdaki kodu **çalıştırmadan** çalışma kâğıdındaki iz tablosunu eşinle doldur.
2. **Python Tutor'da kontrol et:** Kodu aşağıdaki bağlantıyla aç. Biriniz bir sonraki `toplam` değerini söylesin, diğeriniz **Next**'e bassın. Tablonuz farklıysa düzeltin.
3. **Tartış:** Çalışma kâğıdındaki üç soruyu eşinle cevaplayın.

[▶ Bu kodu Python Tutor'da adım adım çalıştır](https://pythontutor.com/visualize.html#code=toplam%20%3D%200%0Afor%20sayi%20in%20range%281%2C%206%29%3A%0A%20%20%20%20toplam%20%3D%20toplam%20%2B%20sayi%0A%20%20%20%20print%28sayi%2C%20toplam%29%0Aprint%28%22Sonu%C3%A7%3A%22%2C%20toplam%29&cumulative=false&curInstr=0&heapPrimitives=nevernest&mode=display&py=311&rawInputLstJSON=%5B%5D&textReferences=false)

```python
toplam = 0
for sayi in range(1, 6):
    toplam = toplam + sayi
    print(sayi, toplam)
print("Sonuç:", toplam)
```

Rolleri değiştirin: klavye eşine geçsin. Yukarıdaki **Adım 3** hücresini değiştirerek görevleri sırayla yap; her görevden sonra sonucu kâğıda yaz.
